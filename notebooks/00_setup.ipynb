# 00 - Configuração
Cria objetos do Unity Catalog e caminhos do MVP. Ajuste os widgets antes da primeira execução.

```python
dbutils.widgets.text("catalog", "workspace", "Catálogo")
dbutils.widgets.text("schema", "mvp_b3", "Schema")
dbutils.widgets.text("volume", "landing", "Volume")
catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")
volume = dbutils.widgets.get("volume")

for value, label in [(catalog, "catalog"), (schema, "schema"), (volume, "volume")]:
    if not value.replace("_", "").isalnum():
        raise ValueError(f"Nome inválido para {label}: {value}")
```

```python
spark.sql(f"CREATE SCHEMA IF NOT EXISTS `{catalog}`.`{schema}` COMMENT 'MVP acadêmico de ações da B3'")
spark.sql(f"CREATE VOLUME IF NOT EXISTS `{catalog}`.`{schema}`.`{volume}` COMMENT 'Arquivos de entrada não versionados'")
spark.sql(f"USE CATALOG `{catalog}`")
spark.sql(f"USE SCHEMA `{schema}`")

base_path = f"/Volumes/{catalog}/{schema}/{volume}"
for folder in ["cotahist", "referencia", "quarentena"]:
    dbutils.fs.mkdirs(f"{base_path}/{folder}")

print(f"Envie COTAHIST_A2025.TXT para: {base_path}/cotahist/")
print(f"Envie setores_b3.csv para: {base_path}/referencia/")
```
